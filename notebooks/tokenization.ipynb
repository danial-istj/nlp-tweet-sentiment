```python
import pandas as pd
import os

# Path to the downloaded raw training dataset
data_path = 'data/raw/train.csv'

# Load the dataset if it exists, otherwise download it
if os.path.exists(data_path):
    data = pd.read_csv(data_path)

else:
    print(f"File '{data_path}' not found. Downloading dataset...")

    from datasets import load_dataset

    dataset = load_dataset('cardiffnlp/tweet_eval', 'sentiment')

    data = dataset['train'].to_pandas()

    # Map numerical labels to human-readable names
    labels_map = {
        0: 'negative',
        1: 'neutral',
        2: 'positive'
    }

    data['label_name'] = data['label'].map(labels_map)

# Create a copy of the dataset
df = data.copy()

# Display dataset information
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nDataset columns:")
print(df.columns.tolist())

# Display first five rows
display(df.head())
```
